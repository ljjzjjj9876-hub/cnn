## 代码维度简单对比

1. 图像分类 CIFAR10
输入：`[B,3,32,32]`
输出：`[B,10]` → 整张图 10 类打分
2. 语义分割（假设 10 类，图片 32×32）
输入：`[B,3,32,32]`
输出：`[B,10,32,32]` → **32×32 每一个像素，都有 10 个类别预测分数**

✅ **图像分类（你现在学的 ResNet18 做的任务）**
输入一张图 → 输出**一整个图片是什么**。
例：图片里有一只猫，模型只输出：这张图是「猫」。

> 
> 只知道**整张图的类别**，不知道猫在图片哪个位置、轮廓在哪。

✅ **语义分割 Semantic Segmentation**
输入一张图 → **图片里每一个像素点，都判断它属于哪一类**。
例：一张图里面有猫 + 草地

- 属于猫的全部像素，标记成「猫」
- 草地的像素标记成「草地」
- 背景像素标记成「背景」
